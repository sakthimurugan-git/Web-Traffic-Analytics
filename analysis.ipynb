```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')
```

```python
df = pd.read_csv('../data/web_traffic.csv')

df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
total_page_views = df['Page Views'].sum()

print("Total Page Views:", total_page_views)
```

```python
avg_session = df['Session Duration'].mean()

print("Average Session Duration:", avg_session)
```

```python
avg_bounce = df['Bounce Rate'].mean()

print("Average Bounce Rate:", avg_bounce)
```

```python
avg_conversion = df['Conversion Rate'].mean()

print("Average Conversion Rate:", avg_conversion)
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Page Views'], bins=20)

plt.title('Distribution of Page Views')
plt.xlabel('Page Views')
plt.ylabel('Frequency')

plt.savefig('../images/page_views.png')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df['Session Duration'])

plt.title('Session Duration Analysis')

plt.savefig('../images/session_duration.png')

plt.show()
```

```python
traffic = df['Traffic Source'].value_counts()

traffic.plot(kind='bar')

plt.title('Traffic Sources')
plt.xlabel('Traffic Source')
plt.ylabel('Number of Users')

plt.savefig('../images/traffic_sources.png')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Conversion Rate'], bins=20)

plt.title('Conversion Rate Distribution')

plt.savefig('../images/conversion_rate.png')

plt.show()
```

```python
plt.figure(figsize=(10,6))

correlation = df.corr(numeric_only=True)

sns.heatmap(correlation, annot=True, cmap='coolwarm')

plt.title('Correlation Heatmap')

plt.savefig('../images/correlation_heatmap.png')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x='Traffic Source',
    y='Conversion Rate',
    data=df
)

plt.title('Traffic Source vs Conversion Rate')

plt.savefig('../images/source_vs_conversion.png')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Time on Page'], bins=20)

plt.title('Time on Page Distribution')

plt.savefig('../images/time_on_page.png')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(x='Previous Visits', data=df)

plt.title('Previous Visits Count')

plt.savefig('../images/previous_visits.png')

plt.show()
```

# Key Insights

1. Website traffic shows varying user engagement patterns.

2. Certain traffic sources generate higher conversion rates.

3. Session duration indicates moderate engagement among users.

4. Bounce rate analysis suggests some users leave the website quickly.

5. Time spent on page has a positive relationship with conversions.

6. Returning visitors show better engagement compared to new visitors.

7. Correlation analysis helps identify factors affecting user behavior and conversions.
